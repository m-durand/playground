```python
from pydub import AudioSegment
from IPython.display import Audio
import os
```

```python
# Read in the file
audio = AudioSegment.from_file("turista_chords.mpeg")
```

```python
# Interactive player
audio.export("preview_original.wav", format="wav")
Audio("preview_original.wav")
```

```python
# parameters 
# len(audio) is 55711 ms
# the sample includes an intro and an outro so I need to play with the start and end times of verses
# also the song has nice gaps so

verses = [
    {"start_ms": 0,    "end_ms": 4500, "gap_after_ms": 1000},  # 1st verse En mi vida fuiste turista ends around 0:49 ok
    {"start_ms": 0,  "end_ms": 22000,  "gap_after_ms": 0},  # 2nd verse Una foto bonita should finish around 1:24 ok until now
    {"start_ms": 0,  "end_ms": 4500,  "gap_after_ms": 4000},  # 3rd verse Que lleva así, lleva así mucho tiempo should end around 2:16 before the gap Jamás y nunca como tú, uh-uh
    {"start_ms": 15000,  "end_ms": 0,    "gap_after_ms": 0},    # 4th verse, keep ending
]
```

```python
# build the loop yeah loopy loop. Going to do this by ear and then adjust parameters
output = AudioSegment.empty()

for v in verses:
    start_ms = v["start_ms"]
    end_ms = v["end_ms"]
    gap_after_ms = v["gap_after_ms"]

    if end_ms == 0:
        verse_audio = audio[start_ms:]
    else:
        verse_audio = audio[start_ms:-end_ms]

    output += verse_audio

    if gap_after_ms > 0:
        output += AudioSegment.silent(duration=gap_after_ms)
```

```python
# Interactive player again
#output.export("preview_looped.wav", format="wav")
Audio("preview_looped.wav")
```
